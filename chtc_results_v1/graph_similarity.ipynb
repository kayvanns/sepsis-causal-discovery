```python
import pickle
graphs = ["FCI_fisherz_simple_a05_10k_d4.pkl", "FCI_fisherz_simple_a05_10k_d-1.pkl"] 
matrices = {}
for file in graphs:
    with open(f"./pkl/{file}", "rb") as f:
            graph, cols = pickle.load(f)
            matrices[file] = graph.graph
            
```

```python
def pag_skeleton_shd(mat1, mat2):

    n = mat1.shape[0]

    shd = 0

    for i in range(n):

        for j in range(i + 1, n):

            edge1 = mat1[i, j] != 0 or mat1[j, i] != 0

            edge2 = mat2[i, j] != 0 or mat2[j, i] != 0

            if edge1 != edge2:

                shd += 1

    return shd

pag_skeleton_shd(matrices["FCI_fisherz_simple_a05_10k_d4.pkl"], matrices["FCI_fisherz_simple_a05_10k_d-1.pkl"])
```

```python
def pag_pairwise_mismatch(mat1, mat2):

    n = mat1.shape[0]

    diff = 0

    for i in range(n):

        for j in range(i + 1, n):

            if (mat1[i, j], mat1[j, i]) != (mat2[i, j], mat2[j, i]):

                diff += 1

    return diff
pag_pairwise_mismatch(matrices["FCI_fisherz_simple_a05_10k_d4.pkl"], matrices["FCI_fisherz_simple_a05_10k_d-1.pkl"])
```
